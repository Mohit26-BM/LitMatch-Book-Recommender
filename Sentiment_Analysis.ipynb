```python
emotion_labels = [
    "admiration",
    "amusement",
    "anger",
    "annoyance",
    "approval",
    "caring",
    "confusion",
    "curiosity",
    "desire",
    "disappointment",
    "disapproval",
    "disgust",
    "embarrassment",
    "excitement",
    "fear",
    "gratitude",
    "grief",
    "joy",
    "love",
    "nervousness",
    "optimism",
    "pride",
    "realization",
    "relief",
    "remorse",
    "sadness",
    "surprise",
    "neutral",
]
isbn = []

emotion_scores = {label: [] for label in emotion_labels}


def calculate_max_emotion_scores(predictions):
    per_emotion_scores = {label: [] for label in emotion_labels}

    # predictions = list of lists of dicts
    for pred in predictions:
        for emo in pred:
            label = emo["label"]
            score = emo["score"]
            per_emotion_scores[label].append(score)

    return {
        label: (max(scores) if scores else 0.0)
        for label, scores in per_emotion_scores.items()
    }
```

```python
from nltk.tokenize import sent_tokenize

for i in range(10):
    isbn.append(books["isbn13"][i])

    # safer sentence splitting
    sentences = sent_tokenize(books["description"][i])
    sentences = [s for s in sentences if len(s.split()) > 3]

    # if no valid sentences, default to neutral
    if not sentences:
        for label in emotion_labels:
            emotion_scores[label].append(0.0)
        emotion_scores["neutral"][-1] = 1.0
        continue

    predictions = classifier(sentences)
    max_scores = calculate_max_emotion_scores(predictions)

    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])
```

```python
from tqdm import tqdm
from nltk.tokenize import sent_tokenize

emotion_labels = [
    "admiration",
    "amusement",
    "anger",
    "annoyance",
    "approval",
    "caring",
    "confusion",
    "curiosity",
    "desire",
    "disappointment",
    "disapproval",
    "disgust",
    "embarrassment",
    "excitement",
    "fear",
    "gratitude",
    "grief",
    "joy",
    "love",
    "nervousness",
    "optimism",
    "pride",
    "realization",
    "relief",
    "remorse",
    "sadness",
    "surprise",
    "neutral",
]

isbn_list = []
emotion_scores = {label: [] for label in emotion_labels}

for i in tqdm(range(len(books))):
    isbn_list.append(books["isbn13"][i])

    # Correct sentence tokenization
    sentences = sent_tokenize(books["description"][i])

    # Get predictions per sentence
    predictions = [classifier(sent)[0] for sent in sentences]

    # Aggregate scores
    max_scores = calculate_max_emotion_scores(predictions)

    # Store for each emotion
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])
```

```python
emotions_df = pd.DataFrame(emotion_scores)
emotions_df["isbn13"] = isbn_list
```
